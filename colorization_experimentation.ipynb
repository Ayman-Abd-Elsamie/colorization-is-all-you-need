### First, let's start with conversions and metrics

```python
import cv2
import numpy as np
```

```python
class Conversions:
    @staticmethod
    def rgb_to_hsv(img):
        return cv2.cvtColor(img, cv2.COLOR_RGB2HSV).astype(np.float32)
    
    @staticmethod
    def rgb_to_lab(img):
        return cv2.cvtColor(img, cv2.COLOR_RGB2LAB).astype(np.float32)
```

```python
class Metrics:

    @staticmethod
    def calc_mean_saturation(img):
        sat_mean = img[..., 1].mean() / 255.0
        return sat_mean
    
    @staticmethod
    def ab_mean_calculation(pred_img, gt_img):
        p = Conversions.rgb_to_lab(pred_img)[..., 1:3]
        g = Conversions.rgb_to_lab(gt_img)[..., 1:3]
        mse = np.mean((p - g) ** 2)
        return mse
```
